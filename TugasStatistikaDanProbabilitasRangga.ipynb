```python
import pandas as pd

path = "https://raw.githubusercontent.com/rangga-ti/TugasStatistikaDanProbabilitas-Rangga/main/rangga.xlsx"
dataraw = pd.read_excel(path)

print(dataraw)
```

```python
datafrq = pd.crosstab(index=dataraw["GRADE"], columns="Frekuensi")

print(datafrq)
```

```python
# Grafik Garis
import matplotlib.pyplot as plt

datafrq.plot(marker='o', color='#0047AB', linewidth=3, markersize=8, legend=False)
plt.title("Grafik Garis Frekuensi Grade", fontsize=12, fontweight='bold')
plt.ylabel("Frekuensi", fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Grafik Batang
datafrq.plot(kind='bar', color='#FF6B6B', edgecolor='black', legend=False)
plt.title("Grafik Batang Frekuensi Grade", fontsize=12, fontweight='bold')
plt.ylabel("Frekuensi", fontweight='bold')
plt.xticks(rotation=0, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Pie Chart
colors = ['#FF4B4B', '#FF8F00', '#00C853', '#00E5FF', '#AA00FF', '#607D8B']
explode = [0.05] * len(datafrq)
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False, colors=colors, explode=explode, startangle=90)
plt.title("Persentase Grade Mahasiswa", fontsize=12, fontweight='bold')
plt.ylabel("")
plt.show()
```

```python
dataraw["NILAI"] = pd.to_numeric(dataraw["NILAI"], errors='coerce')
dt = dataraw["NILAI"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()

print(stats)
```
